## 1.) Null Polarimetry

The null in the transmission is found with $\theta_{QWP} = 30^{\circ}$ and $\theta_{LA} = 60^{\circ}$ with respect to $x$. The angle for which a null is found must have the incident light be perpendicular to the linear polarizer such that:

$$\theta_{LA, null} = 60^{\circ} + 90^{\circ} = 150^{\circ}$$

Working backwards to find the angle that this polarized light (after going through the QWP) must be with respect to the FA of the QWP we get:

$$\alpha = \theta_{LA, null} - \theta_{QWP} = 120^{\circ}$$

The Jones vector for this polarization in between the two elements can be described with:

$$E_{between} = \begin{pmatrix} \cos \alpha \\ \sin \alpha \end{pmatrix} = \begin{pmatrix} -\frac{1}{2} \\ -\frac{\sqrt{3}}{2} \end{pmatrix}$$

We can then use the Jones matrix for the QWP to work backward to get the initial polarization $E_0$:

$$E_{0} = \begin{pmatrix} 1 & 0 \\ 0 & -i \end{pmatrix} E_{between} = \begin{pmatrix} -\frac{1}{2} \\ i\frac{\sqrt{3}}{2} \end{pmatrix}$$

The ellipticity can be defined as the ratio of the minor axis $(b)$ and the major axis $(a)$:

$$\epsilon = \frac{b}{a} = \left|\frac{\frac{-1}{2}} {i \frac{\sqrt{3}}{2}} \right| = \frac{1}{\sqrt{3}}$$

The ellipticity angle is found with:

$$\chi = \tan^{-1} \left| \frac{b}{a} \right| = 30^{\circ}$$

The orientation angle has been found when we rotated the frame to match the polarization and is:

$$\psi = \alpha = 120^{\circ}$$

## 2.) Optoisolator

### 2a.)

The system can be treated as a series of matrices representing the light propagating forward through the optical elements and then in reverse after it hits the mirror.

$$\underbar{M}_{final} = \underbar{M}_{HP,B} \underbar{M}_{QWP,B} \underbar{M}_{mirror} \underbar{M}_{QWP, F} \underbar{M}_{HP,F}$$

Assuming the light enters the system unpolarized, the first element it interacts with is the horizontal polarizer $(\underbar{M}_{HP})$ with a matrix represented by:

$$\underbar{M}_{HP} = \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix}$$

With the matrix for the quarter waveplate $(\underbar{M}_{QWP})$ represented as:

$$\underbar{M}_{QWP} = e^{i\pi/4} \begin{pmatrix} 1 & 0 \\ 0 & -i \end{pmatrix}$$

But since it is rotated with its fast axis at $+45^{\circ}$ we must use the rotation matrices.

$$\underbar{M}_{QWP,+45^{\circ}} = \underbar{R}(-\theta) \underbar{M}_{QWP} \underbar{R}(\theta)$$

$$\underbar{M}_{QWP,+45^{\circ}} = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ -1 & 1 \end{pmatrix} e^{i\pi/4}  \begin{pmatrix} 1 & 0 \\ 0 & -i \end{pmatrix} \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & -1 \\ 1 & 1 \end{pmatrix}$$

$$\underbar{M}_{QWP,+45^{\circ}} = \frac{1}{2} e^{i\pi/4} \begin{pmatrix} 1-i & 1+i \\ 1+i & 1-i \end{pmatrix}$$

$$\underbar{M}_{QWP,+45^{\circ}} = \frac{1}{2} e^{i\pi/4} (1-i) \begin{pmatrix} 1 & i \\ i & 1 \end{pmatrix}$$

$$\underbar{M}_{QWP,+45^{\circ}} = \frac{1}{2} \left[\frac{1+i}{\sqrt{2}} \right] (1-i) \begin{pmatrix} 1 & i \\ i & 1 \end{pmatrix}$$

$$\underbar{M}_{QWP,+45^{\circ}} = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & i \\ i & 1 \end{pmatrix}$$

For the mirror $(\underbar{M}_{mirror})$, the matrix is:

$$\underbar{M}_{mirror} = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$$

On the return trip the QWP is now at $-45^{\circ}$ which gives:

$$\underbar{M}_{QWP,-45^{\circ}} = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & -i \\ -i & 1 \end{pmatrix}$$

Combining this all together:

$$\underbar{M}_{final} = \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & -i \\ -i & 1 \end{pmatrix} \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & i \\ i & 1 \end{pmatrix} \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix}$$

```julia
m_qwp_b = [1 -im ; -im 1]
m_hp = [1 0 ; 0 0]
m_qwp_f = [1 im ; im 1]
m_mirror = [1 0 ; 0 -1]

m_f = m_hp * (1/sqrt(2)) * m_qwp_b * m_mirror * (1/sqrt(2)) * m_qwp_f * m_hp

m_before = (1/sqrt(2)) * m_qwp_b * m_mirror * (1/sqrt(2)) * m_qwp_f * m_hp

println(m_f)
println(m_before)
```
